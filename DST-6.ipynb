```python
import pandas as pd
from pandas.api.types import is_numeric_dtype

# Fill missing values column by column
for col in df.columns:
    if is_numeric_dtype(df[col]):
        # Numeric columns: fill with median
        median_value = df[col].median()
        if pd.notna(median_value):
            df[col] = df[col].fillna(median_value)
    else:
        # Text/categorical columns: fill with mode
        mode_value = df[col].mode()
        fill_value = mode_value.iloc[0] if not mode_value.empty else "Unknown"
        df[col] = df[col].fillna(fill_value)

# Check missing values again
print(df.isnull().sum())
```

```python
# Remove duplicates
df.drop_duplicates(inplace=True)

# Convert Crop_Year into date format
df["Crop_Year"] = pd.to_datetime(
    df["Crop_Year"].astype(str) + "-01-01"
)

print(df.dtypes)
print(df.head())
```

```python
# loc - select rows using index labels
print(df.loc[0:4, ["Crop", "Season", "Yield"]])

# iloc - select rows and columns using positions
print(df.iloc[0:5, 0:4])

# Select a single column
print(df.loc[:, "Crop"])

# Select first 5 rows
print(df.iloc[:5])
```

```python
# Filter rows where Yield > 1
print(df[df["Yield"] > 1])

# Filter rows where Crop is Coconut
print(df[df["Crop"] == "Coconut"])

# Multiple conditions
print(df[(df["Yield"] > 1) & (df["Area"] > 1000)])
```

```python
# Create a new column
df["Production_Per_Area"] = df["Production"] / df["Area"]

# Display selected columns
print(df[["Crop", "Production", "Area", "Production_Per_Area"]].head())
```

```python
from sklearn.preprocessing import OrdinalEncoder

# Create ordered year categories
df["Year_Group"] = pd.cut(
    df["Crop_Year"].dt.year,
    bins=[1996, 2005, 2015, 2020],
    labels=["Old", "Middle", "Recent"]
)

# Ordinal Encoding
oe = OrdinalEncoder(
    categories=[["Old", "Middle", "Recent"]]
)

df["Year_Encoded"] = oe.fit_transform(
    df[["Year_Group"]]
).astype(int)

print(df[["Crop_Year", "Year_Group", "Year_Encoded"]].head())
```
